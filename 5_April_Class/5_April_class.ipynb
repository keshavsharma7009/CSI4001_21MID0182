```python
import re
import nltk
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.feature_extraction.text import TfidfVectorizer
from tqdm import tqdm
import gensim.downloader as api
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
df = pd.read_csv("IMDB Dataset.csv", engine='python', on_bad_lines='skip')
```

```python
df['sentiment'] = df['sentiment'].map({'positive': 1, 'negative': 0})
```

```python
def clean_text(text):
    text = re.sub(r'<.*?>', '', text) 
    text = re.sub(r'[^a-zA-Z]', ' ', text)
    tokens = nltk.word_tokenize(text.lower())
    return ' '.join([lemmatizer.lemmatize(word) for word in tokens if word not in stop_words])
```

```python
# Downloading all tokenization-related resources
nltk.download('punkt')
nltk.download('punkt_tab')  # Checking if punkt_tab is part of our environment
```

```python
df['clean_review'] = df['review'].apply(clean_text)
```

```python
# --- 2. TF-IDF + Logistic Regression ---
tfidf = TfidfVectorizer(max_features=5000)
X_tfidf = tfidf.fit_transform(df['clean_review'])
X_train, X_test, y_train, y_test = train_test_split(X_tfidf, df['sentiment'], test_size=0.2, random_state=42)
```

```python
model1 = LogisticRegression()
model1.fit(X_train, y_train)
preds1 = model1.predict(X_test)
```

```python
print("TF-IDF + Logistic Regression")
print("Accuracy:", accuracy_score(y_test, preds1))
print("Confusion Matrix:\n", confusion_matrix(y_test, preds1))
```
