```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vect = CountVectorizer(binary = True)
corpus = ["Tessaract is good optical character recognition engine", "optical character recognition is significant"]
vect.fit(corpus)
```

```python
vocab = vect.vocabulary_ #not a function an attribute
for key in sorted(vocab.keys()):
    print("{}:{}".format(key,vocab[key]))
```

```python
print(vect.transform(["This is a good optical illusion. It is a good one"]).toarray())
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vect.transform(["Google Cloud Vision is a character recognition engine"]))
print(similarity)

print(vect.transform(["Google Cloud Vision is a character recognition engine"]).toarray())
```

```python
from nltk.corpus import names
import random
```

```python
# labeled_names = [(names,'male') for name in names.words('male.txt')] + [(names,'female') for name in names.words('female.txt')]


labeled_names = [(name, 'male') for name in names.words('male.txt')] + [(name, 'female') for name in names.words('female.txt')]
```

```python
import random 
random.shuffle(labeled_names)
```

```python
# featuresets = [(gender_features(n), gender) for (n,gender) in labeled_names]

featuresets = [(gender_features(n), gender) for (n, gender) in labeled_names]
```

```python
train_set, test_set = featuresets[5000:], featuresets[:2000]
```

```python
# import nltk
# classifer = nltk.NaiveBayesClassifer.train(train_set)
# classifier.classify(gender_features('David'))
# print(nltk.classify.accuracy(classifier,test_set))

import nltk

# Ensure you've already defined `train_set` and `test_set` before this

# Correct the typo in NaiveBayesClassifier
classifier = nltk.NaiveBayesClassifier.train(train_set)

# Test classification
print(classifier.classify(gender_features('David')))

# Check accuracy
print(nltk.classify.accuracy(classifier, test_set))
```
